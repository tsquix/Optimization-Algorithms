```python
import random

N = 100
TASKS = [round(random.uniform(10, 90), 2) for _ in range(N)]
CPU_SPEEDS = [1.0, 1.25, 1.5, 1.75]
NUM_CPUS = 4

def calculate_time(genome: list):
    cpu_times = [0.0] * NUM_CPUS

    for task_idx, cpu in enumerate(genome):
        cpu_times[cpu] += TASKS[task_idx] / CPU_SPEEDS[cpu]

    return max(cpu_times)

def mutate(genome: list):
    new_genome = list(genome)
    task_idx = random.randint(0, N - 1)

    old_cpu = new_genome[task_idx]
    new_cpu = random.randint(0, NUM_CPUS - 1)

    while new_cpu == old_cpu:
        new_cpu = random.randint(0, NUM_CPUS - 1)

    new_genome[task_idx] = new_cpu
    return new_genome

def evolve(iteration_count=100000):
    parent = [random.randint(0, NUM_CPUS - 1) for _ in range(N)]
    parent_cost = calculate_time(parent)

    for _ in range(iteration_count):
        child = mutate(parent)
        child_cost = calculate_time(child)

        if child_cost <= parent_cost:
            parent = child
            parent_cost = child_cost

    return parent_cost, parent

best_global_cost = float('inf')
best_global_genome = []

print("Start: CPU Optimization\n")

for attempt in range(1, 11):
    cost, genome = evolve(100000)
    print(f"Round {attempt}/10: Shortest time found = {round(cost, 2)}")

    if cost < best_global_cost:
        best_global_cost = cost
        best_global_genome = genome

print("\n" + "=" * 40)
print(f" Final optimal execution time = {round(best_global_cost, 2)}")
```
