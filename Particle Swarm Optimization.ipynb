```python
import random
import math

particle_count = 100
max_vel = 0.03
epoch_count = 50000

W = 0.7   # Inertia
C1 = 1.5  # Cognitive component
C2 = 1.5  # Social component

particle_list = []
global_best_score = -1.0
global_best_cords = (0.0, 0.0)

def calc_score(x, y):
    return abs(math.sin(x) + math.sin(2*x) + math.sin(4*x) + math.sin(8*x)) + \
           abs(math.cos(y) + math.cos(2*y) + math.cos(4*y) + math.cos(8*y))

for _ in range(particle_count):
    rand_x = random.uniform(0, 2 * math.pi)
    rand_y = random.uniform(0, 2 * math.pi)
    rand_v_x = random.uniform(-max_vel, max_vel)
    rand_v_y = random.uniform(-max_vel, max_vel)

    score = calc_score(rand_x, rand_y)

    particle = {
        'cords': (rand_x, rand_y),
        'velocity': (rand_v_x, rand_v_y),
        'pb_cords': (rand_x, rand_y),
        'pb_score': score
    }

    particle_list.append(particle)

    if score > global_best_score:
        global_best_score = score
        global_best_cords = (rand_x, rand_y)

print(f"Start! Best initial guess: {round(global_best_score, 4)}\n")

for epoch in range(1, epoch_count + 1):

    for particle in particle_list:
        x, y = particle['cords']
        v_x, v_y = particle['velocity']
        pb_x, pb_y = particle['pb_cords']
        gb_x, gb_y = global_best_cords

        r1_x, r1_y = random.random(), random.random()
        r2_x, r2_y = random.random(), random.random()

        new_v_x = (W * v_x) + (C1 * r1_x * (pb_x - x)) + (C2 * r2_x * (gb_x - x))
        new_v_y = (W * v_y) + (C1 * r1_y * (pb_y - y)) + (C2 * r2_y * (gb_y - y))

        new_v_x = max(-max_vel, min(max_vel, new_v_x))
        new_v_y = max(-max_vel, min(max_vel, new_v_y))

        new_x = (x + new_v_x) % (2 * math.pi)
        new_y = (y + new_v_y) % (2 * math.pi)

        particle['velocity'] = (new_v_x, new_v_y)
        particle['cords'] = (new_x, new_y)

        new_score = calc_score(new_x, new_y)

        if new_score > particle['pb_score']:
            particle['pb_score'] = new_score
            particle['pb_cords'] = (new_x, new_y)

            if new_score > global_best_score:
                global_best_score = new_score
                global_best_cords = (new_x, new_y)

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}/{epoch_count} -> Current Global Maximum: {round(global_best_score, 5)}")

print("\n" + "="*50)
print("END")
print(f"Highest maximum found: {round(global_best_score, 5)}")
print(f"Located at coordinates: X = {global_best_cords[0]:.5f}, Y = {global_best_cords[1]:.5f}")
print("="*50)
```
